## 1.
გადმოგვეცემა რიცხვების ლისტი, ჩვენ უნდა ვიპოვოთ
საშუალო, მოდა, მედიანა, სტანდარტული გადახრა და დავაბრუნოთ dictionary-ს სახით.

````python
def func(vec: list) -> dict:
  vec.sort()
  res = {}
  ## median
  res["median"] = (vec[len(vec) // 2] + vec[(len(vec) - 1) // 2]) / 2
  ## mean
  freq = {}
  sum = 0
  standart = 0

  for items in vec:
    freq[items] = freq.get(items, 0) + 1
    sum += items

  avg = float(sum / len(vec))
  res["mean"] = avg
  ## standart
  ## due to the population, we dont need Bessel's correction
  for items in vec:
    standart += (items - avg) ** 2
  standart = (standart / len(vec)) ** (1/2)
  res["standart_deviation"] = standart
  ## mode
  loc_max = 0
  vals = []
  for x, y in freq.items():
    if y > loc_max:
      loc_max = y
      vals = [x]
    elif y == loc_max:
      vals.append(x)
    res["mode"] = vals
  ## return
  return res

vec = [1, 3, 5, 29, 9, 1438, 3]
print(func(vec))
````

გადმოგვეცემა string-ების სია და ჩვენ უნდა დავაბრუნოთ dictionary dict[str, list[str]],
სადაც გასაღებად იქნება თითოეული სიტყვის ასოების სორტირებული ვერსია,
ხოლო მნიშვნელობად — იმ სიტყვების სია, რომლებიც ერთმანეთის ანაგრამებია (ანუ შედგება იგივე ასოებით სხვადასხვა თანმიმდევრობით).

words = ["listen", "silent", "enlist", "ratt", "tar", "art", "hello", "below", "elbow", "rat"] ->

{'eilnst': ['listen', 'silent', 'enlist'],
 'artt': ['ratt'],
 'art': ['tar', 'art', 'rat'],
 'ehllo': ['hello'],
 'below': ['below', 'elbow']}

````python
def dic(words: list) -> dict:
  res = {}
  for items in words:
    s = "".join(sorted(items))
    if s not in res:
      res[s] = []
    if items not in res[s]:
      res[s].append(items)
  return res


words = ["listen", "silent", "enlist", "ratt", "tar", "art", "hello", "below", "elbow", "rat"]
dic(words)
##print(dic(words))
````

```
[[1],
 [1, 4],
 [1, 4, 9],
 [1, 4, 9, 16],

 [1, 4, 9, 16, 25, 36],
 [1, 4, 9, 16, 25, 36, 49],

 [1, 4, 9, 16, 25, 36, 49, 64, 81],
 [1, 4, 9, 16, 25, 36, 49, 64, 81, 100]]
```

````python
def func(n: int) -> list:
  l = [];
  for i in range(1, n+1):
    cur = []
    for j in range(1, i+1):
      cur.append(j**2)
    l.append(cur)
  return l

print(func(10))
````

გადმოგვეცემა სორტირებული ლისტი და რაღაც რიცხვი. ჩვენი მიზანია ეს რიცხვი ისე ჩავამატოთ ლისტში რომ სორტირება არ დაირღვეს.

````python
l = [1, 3, 5, 7, 9, 11, 15, 24, 199]
n = int(input())

low, high = 0, len(l) - 1

while low <= high:
  mid = (low+high) // 2
  if l[mid] <= n:
    low = mid + 1
  else:
    high = mid - 1

l.insert(low, n)

print(l)
````
